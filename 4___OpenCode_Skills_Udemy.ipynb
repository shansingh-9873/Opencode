# 1.. What are Skills : Real - World Scenario ?


````python
'''

Hi there 👋 , 

In this section , we are going to deep dive into something called : Skills 🚀

        

So Before understanding what Skills are , first lets understand WHY Skills are important ?
    

So lets break this down with simple examples :

👉 See , LLMs like Gemini , GPT Models , Claude & Even Free Models behind Open Code— 
    They are GENERAL PURPOSE Language Models !
    
    What does General Purpose mean ?
        👉 It means they have the capability of writing , coding , reasoning , 
           translating , summarizing — across various domains !
        👉 They are like a Jack of all Trades !



But here's the PROBLEM 🚨 :

    👉 These LLMs are GOOD at general capabilities ..
    👉 But NOT for specific-task types !

    These models LACK , when u ask them to execute SPECIAL tasks !
    & thats where u can use Skills to execute special tasks !
    
    
    Skills are primarily used : when u want to execute specialised tasks + task is repeated + u want consistent results !
    
    
    


Lets understand it using a various real-world example 🎯 :

    Example 1 : [show life-cycle] : 
                See in any Data Analytics Projects , first thing u have to do is :
                u have to Extract/Collect data !
                
                Imagine you have extracted data from APIs or some 3rd Party Sources like Github :
                [Show GITHUB data]
                
                
                [show life-cycle] : 
                Now next thing is u have to do is :
                u have to clean it , so that u can create final dashboard & show it to Manager !
                
                
                
    Why skills ?
        Bcz Company may have own guidelines regarding Data cleaning ! ( so its a specialised task)
        See Once data is fetched , we have to clean it on daily basis to do analysis bcz Sales Keeps on updated on daily basis
            ( so data cleaning becomes a repeated task)
        
        All we want is : 
            We want to follow some data cleaning rules like :  
            
            1. Remove duplicate & blank rows from sales data
            2. Remove malformed rows in the `Order Date` column. 
            3. Parse `Order Date` into a proper datetime type.
            4. Create `total sales` = `Quantity Ordered × Price Each`
            5. We want our agent to Handle missing or malformed values safely and report what was cleaned.

    

    So whenever you want to do some task AGAIN & AGAIN with :
            👉 Reliably 
            👉 + Repeatedly 
            👉 + with Specialised prompts

    That's where the IDEA of SKILLS comes ✨ ..




'''
````

````python
'''


Now lets understand the STRUCTURE of Skills 📂 :
[show OpenCode Skills Page]

Even u go to opencode skills segment , u can see file structure of skills !

Project config: .opencode/skills/<name>/SKILL.md
Global config: ~/.config/opencode/skills/<name>/SKILL.md
Project Claude-compatible: .claude/skills/<name>/SKILL.md
Global Claude-compatible: ~/.claude/skills/<name>/SKILL.md
Project agent-compatible: .agents/skills/<name>/SKILL.md
Global agent-compatible: ~/.agents/skills/<name>/SKILL.md





Here is the basic hierarchy :

    📁 Project folder
        └── 📁 .openCode folder
              └── 📁 skills folder
                    └── 📁 skill folder
                          └── 📄 SKILL.md   ← (Very important file !)



👉 Inside the skill folder , we have a file called SKILL.md
    

👉 Inside SKILL.md file , ALL the detailed instructions have been written 
   to perform specialised + repeated task !

    eg : For example, imagine that every day you need to:

        Extract Updated Sales data from Github
        Clean and process the data
        Generate a report & Submit the report to managers
    
    
Since this is a repeated workflow, creating a Skill provides:

        ✅ Reliability
        ✅ Consistency
        ✅ Specialized instructions that don't need to be re-written every time
    

    In SKILL.MD , u can mention data-cleaning rules : 

        👉 Which API endpoint to call ?
        👉 Which col-names to extract from the response ?
        👉 How to Handle timeout and authentication errors ?
        👉 What about Output format of Data : CSV, Excel, JSON, or SQLite

        
    All these instructions live inside SKILL.md !










And in the next session …
We'll go EVEN DEEPER and understand the internal structure of SKILL.md file 



'''
````

# 2.. Skills Deep-dive : YAML Front-matter + Markdown body


````python
'''

Hi there 👋 , In the last session …
We understood WHAT Skills are & the FOLDER structure of Skills !




So now , lets deep dive into how to create Skills !


Imagine that once u extract data ,  every day you need to:

        [show life-cycle]
            Clean data as per company rules
            Store cleaned data for further usage
            Generate a Sales report & Submit the report to managers
    
    
Since this is a repeated workflow, creating a Skill makes sense !



So before creating Skill , first lets understand components of Skill & then in the next session , 
I will show u practical Part !




Lets understand : 👉 What does the Skill file which is SKILL.md ACTUALLY contains ?


     👉 SKILL.md actually has 2 IMPORTANT things :

        1) YAML Front-matter
        2) Markdown Body


Lets understand each of these one by one 👇




═══════════════════════════════════════════════
1) 📝 YAML Front-matter
═══════════════════════════════════════════════

YAML Front-matter is the TOP section of the SKILL.md file !
It has 2 things :

    👉 a) Skill name
    👉 b) Description



👉 a) Skill name :
            Using skill-name , OPENCODE can IDENTIFY the skill !

👉 b) Description :
            Description is VERY IMPORTANT bcz ..
            After reading the description , OPENCODE knows WHEN to use or load this Skill !







═══════════════════════════════════════════════
2) 📄 Markdown Body
═══════════════════════════════════════════════

    Here we WRITE DOWN EVERYTHING :-
    
    👉 All the instructions
    👉 + Code (if needed)
    👉 + Coding patterns
    👉 + Validation patterns


    + Link All files/references , like if your company have provided Python script for Data cleaning  ,
       so here you can LINK that !






🎯 Lets visualize how a SKILL.md file looks :


```md
┌──────────────────────────────────────────────────────────┐
│  ---                                                     │
│  name: fetch-amazon-data                                 │
│  description: Use this Skill to fetch the six Amazon     │
│               datasets (inventory, products, returns,    │
│               reviews, sellers, shipping) from the       │
│               GitHub resources repo and save them as     │
│               timestamped CSVs with a run log            │
│  ---                                                     │
│                                                          │
│  # 📦 Fetch Amazon Data from API                        │
│                                                          │
│  ## Instructions                                         │
│                                                          │
│                
│  - Fetch all six CSV URLs                │
│  - Save data to a timestamped data/ folder          │
│        (format: YYYY-MM-DD_HH-MM-SS)                     │
│  - Write a run log: URLs called, successes,              │
│        failures, and any errors encountered              │
│                                                          │
│  ## 📚 Files Used                                        │
│                                                          │
│  - scripts/fetch.py                                      │
│        → Async download + save logic                     │
│                                                          │
│  - data/<timestamp>/                                     │
│        → Saved CSV files (one per dataset)               │
│                                                          │
│  - logs/<timestamp>/fetch.log                            │
│        → Log of every API call made                      │
└──────────────────────────────────────────────────────────┘
```






🟢 Quick Summary :

    👉 YAML Front-matter  =  Identity card (name + description)
    👉 Markdown Body      =  The detailed instruction manual

    👉 Name        →  Tells OPENCODE — "Who am I ?"
    👉 Description →  Tells OPENCODE — "WHEN to use me ?"
    👉 Body        →  Tells OPENCODE — "HOW to do the task ?"



& in the next session …
We will understand HOW OPENCODE actually LOADS these Skills internally —
This is where you'll learn about a beautiful concept called : "Progressive Disclosure" 🚀


'''
````
