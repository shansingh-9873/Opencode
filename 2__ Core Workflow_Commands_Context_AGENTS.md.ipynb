```python
'''



[first show this in OpenCode] :

U will see something interesting , this says 10% of your Context have been filled !

This number tells : How many tokens we have used so far in this conversation !








━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
🎙️  NARRATION SCRIPT
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Before we understand the Context Window , we first have to UNDERSTAND what Context is :




Lets connect it with a real-life example !
Imagine YOU are a Software Engineer / App Developer joining a new company .
What is the CONTEXT you need to actually do your job well/ or to your understand project well ?

    a)  Your Codebase 📂
            👉 the functions you have written , the folder structure , the naming conventions ..
            👉 because depending on THIS , you decide what the next line of code should be !

    b)  Your PRD / Spec document 📝
            👉 this is where ALL the requirements are usually written .

    c)  Your Jira / GitHub Issues 🐙
            👉 based on these tools , you get your day-to-day tasks ,
               like a bug or a feature you have to build in your software .

    d)  Slack / Team Messages 💬
            👉 where all your teammates discuss things .

    e)  Previous AI Chats 🤖
            👉 where you have already talked to Codex earlier about this project .

    f)  Git history 🌿
            👉 what changed , when , and why .

👉 ALL of this together is your "context" .


👉 So , what is "Context" ?
    Context is ALL the information needed to understand something correctly ..







👉 So What is a Context Window in OpenCode ?

    The Context Window is the amount of information ( measured in TOKENS )
    that a models(brain of OpenCode) can SEE and REMEMBER at one time ,
    while it generates a response for you ..
                OR
    In simple terms , it is a Memory limitation of a LLM in a particular chat 

    & Token is roughly around 4 characters
    
    
    

    🧠 Think of it like the model's WORKING MEMORY .


Doesn't matter what model u are using in OpenCode !
It could be some free model like DeepSeek variant
            Or paid one  , like :
                            Claude Models
                            OpenAI Models
                            Gemini LLMs


    Every model have diff context window size :
    
    Opus : 1M Tokens
    Sonnet : 200K Tokens
    GPT 5 : 1M Tokens
    
    Big Pickle : 200K Tokens



'''
```

# 4.. Auto-compaction in OpenCode

```python
'''


Hi there 👋 , in the last session ..
We saw every component inside the context window !

Now let's understand : 👉 What is Auto-Compaction , and why does OpenCode Code do it ?


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
🎙️  NARRATION SCRIPT
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━


                         
        [Open VS CODE] : 
        Give this Prompt :
        tell me tokens reserved for auto-compaction for current Model
        

    Around 20k Tokens are reserved for Auto compaction !
 

    Context window usage :
    [████████████████████░░░░]   ← imagine you're at 80–90% of Context Window usage , say ~160K tokens used


So what does OpenCode do to save you ?
        👉 It COMPACTS the context — it summarises the conversation so far into a much
            shorter form & store it into a Auto-compaction buffer !

When OpenCode does this AUTOMATICALLY , we call it AUTO-COMPACTION ..




[VS CODE]  show : 
    Imagine I am using OpenCode 
    my window is ~80–90% full ? 
    [U can see number(80-90%) which tells how much of your Context have been filled !

    How do I handle it ?"

    There are 2 ways to do AUTO-COMPACTION : 


    a)  Let OpenCode handle it AUTOMATICALLY 
            👉 OpenCode summarises conversation & store it into Auto Compaction Buffer & makes your token usage free !
            👉 But remember : it's not loss-less( it means u will loose some essence of info while summarising) 
                    , and OpenCode may trigger LATE .
            
        But when Auto-compaction buffer fills , OpenCode will not allow u to chat, degrades response quality 
        So close that session & create a new one !
        
        

    b)  Handle it MANUALLY with the  /compact  command  ⭐ ( the better habit )
            👉 You decide WHEN to compact , before quality of OpenCode drops .
            👉 run /compact command !
                [ Show /compact via VSCODE :]
            
            
        

'''
```

# 5.. what AGENTS.md is & how to create it (/init)

```python
'''

========================================
PART 1 — Why AGENTS.md Matters
========================================



So before understanding What is AGENTS.md , lets first understand why there is a need of AGENTS.md !


let say : You are a Data Engineer at Amazon & your Job is to make Insights from Amazon Sales Data !

    So You spent 2 hours with OpenCode & built amazing things !


    You have : 

    Collected Amazon sales data from various sources 
    Then [show dashboard] : 
        Performed Data Analysis 
        Built KPIs (Total Revenue, Total Orders , Top-Selling Products)
        Built dashboards so that u can showcase it to Business Team/Managers 
    
    
    
    
    
Everything is working perfectly.
You’re confident.
You’re proud..





Now next day… Your Manager says :
👉 "Can you improve this Amazon Sales Dashboard and add more business insights?"
    eg like : Impact of Festivals on Sales ?
              Analysing Peak hour sales !



Then You open a new AI session in OPENCode !
And you say:
    👉 “Help me improve my Amazon analysis dashboard as I want to add these following Insights”

And suddenly…
Then OpenCode asks:

    👉 "What does your dataset look like?"
    👉 "What columns are available?"
    👉 "How are you calculating Total Revenue?"
    👉 "Which KPIs have already been implemented?"



And you’re like…
😩 “I already explained all of this yesterday!”
    So its Frustrating… right?

[Pause]



👉 Bcz LLMs/Models behind OpenCode don’t have memory across sessions..
    LLMs are memory-less / stateless 
            OR
    OpenCode don’t remember anything across sessions.

They only remember what’s inside the current conversation/session..
And this creates a HUGE problem…






So now the question is…

👉 How do we make OpenCode REMEMBER our project?"
    One way is : Explain things from scratch into every new session ( thats practically impossibe 
                    in bigger projects & this is so error-prone)
                    
    Second way is : You could give OPenCode one file… & that file contains all your
                    Project info 
                    And whenevr u start a new session  , Claude will read that file & understands your entire project instantly 
                    & that file is known as AGENTS.md !
                    





========================================
PART 2 — A Simple One-Line Definition
========================================


AGENTS.md is :

    👉 A plain text/markdown file in your project folder which contains your project's rules, conventions, and warnings.
                                            OR
        AGENTS.md is like a context & context is everything when u are working LLMs like CLAUDE



In simple words —

    👉 README.md  is designed for humans which tells what your project does..
    👉 AGENTS.md  is designed for OpenCode which tells to your OpenCode Agent what your project is all about 


    👉 README.md  =  for humans  ( what your project does )
    👉 AGENTS.md  =  for OpenCode   ( how YOUR project should be worked on )
                    ie it is like a : rule-book + Summary of your Project for OpenCode which tells CLAUOpenCodeDE :
                                                    -> What your Project is :
                                                    -> How your Code is structured :
                                                    -> What Tools & libraries u are going to using 




========================================
PART 3 — Where Does this "AGENTS.md" Live ?
========================================


Super simple.

    👉 It lives in the ROOT of your project folder
    👉 The filename is exactly :   AGENTS.md
    
    [go to https://opencode.ai/docs/rules/]



Example : How folder structure for our project looks like :


    📂 Sales/
        ├── 📜 AGENTS.md            ← this is the magic file
        ├── 📜 README.md             ← for humans
        ├
        ├── 📂 data/
        │     └── amazon_sales_customer.sqlite          
        └── 







OPENCODE sees AGENTS.md in the root → reads it → applies your rules.
Done.

Now in the next session , we will write First AGENTS.md file !



'''
```
