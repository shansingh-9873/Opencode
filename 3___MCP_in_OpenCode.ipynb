# 2..  [LIVE DEMO] : Github MCP - Server ! ( Integrate Github to Claude Code ) 

```python
'''


Let's understand how we can use the **GitHub MCP Server** in a real-world scenario.


[show Github] : 

>   Let say your comapny sales data been updated on daily basis on Github !
    So that we will extract it , process it & make insights out of that !





SO if you ask Opencode :
       
       👉 "Hey OpenCode , go & Fetch Updated Sales Data from Github !


    OpenCode can't do that !
    Because **GitHub is an external service**.

    OpenCode has **no direct access** to your GitHub repositories. 
    It cannot see your files, commits, branches, or repository contents unless you explicitly connect it to GitHub.

    This is exactly where the **GitHub MCP Server** comes in.
    So Once connected, OpenCode can access GitHub on your behalf and perform actions such as:

     you can simply ask OpenCode in plain English:

* **"Has yesterday's Amazon sales data been updated?"**
* **"Show me the latest commit in the sales repository."**
* **"Are there any open issues in my sales repository?"**
* **"Create a pull request for today's changes."**

It means , 
In short:

> **Without the GitHub MCP Server, OpenCode cannot access your GitHub data.**
>
> **With the GitHub MCP Server connected, OpenCode can securely interact with your repositories using natural language.**



'''
```

### Step 1 — Get a Personal Access Token (PAT)

```python
'''

github.com
   → Settings
      → Developer Settings
         → Personal Access Tokens
            → Fine-grained tokens
               → Generate new token
                  → Copy the token
                  
                  
                  
                  
    ⚠️ Copy it immediately — GitHub only shows the token once. If you lose it, you make a new one.
        When generating, give it access to the repositories you want Claude to work with, 
            and grant the relevant permissions (e.g. read/write for Contents, Issues, Pull Requests).





'''
```

### Step 2 — Add the Server to OpenCode

```python
'''


    Google as :
    Github MCP server Opencode 
    
    Go to Github !
    
    
Inside opencode.json file !
Put this :

    
    
{
  "$schema": "https://opencode.ai/config.json",
  "mcp": {
    "github": {
      "type": "remote",
      "url": "https://api.githubcopilot.com/mcp/",
      "enabled": true,
      "oauth": false,
      "headers": {
        "Authorization": "Bearer YOUR_GITHUB_PAT"
      }
    }
  }
}







'''
```

### Step 3 — Verify the Connection

```python
'''

Open a new OpenCode session and run:

/mcps


You should see github listed with a checkmark ✅



'''
```

# 3.. MCP- database-server ( Talk to your DB without writing code) 🗄️


```python
'''
🎯 What is mcp-database-server ?

    👉 DBHub = MCP - Database Server


    Using this MCP Server , you can talk to your database DIRECTLY 
    WITHOUT writing any Python code ! 🤯





DBhub MCP supports SQLite, SQL Server, PostgreSQL, and MySQL databases..



            +------------------+    +--------------+    +------------------+
            |                  |    |              |    |                  |
            |                  |    |              |    |                  |
            |  Claude Desktop  +--->+              +--->+    PostgreSQL    |
            |                  |    |              |    |                  |
            |  Claude Code     +--->+              +--->+    SQL Server    |
            |                  |    |              |    |                  |
            |  Cursor          +--->+    DBHub     +--->+    SQLite        |
            |                  |    |              |    |                  |
            |  VS Code         +--->+              +--->+    MySQL         |
            |                  |    |              |    |                  |
            |  Copilot CLI     +--->+              +--->+    MariaDB       |
            |                  |    |              |    |                  |
            |                  |    |              |    |                  |
            +------------------+    +--------------+    +------------------+
                 MCP Clients           MCP Server             Databases







'''
```

```python
'''

Hi there 👋 , Welcome to one of the MOST IMPORTANT sessions in this entire course ! 🚨

In the last 3 sessions , we got EXCITED about MCP — & how it gives Claude SUPER POWERS !
But with great power , there comes great responsibility 🕷️ ..


🚨 So once you give Open code access to your Browser , your GitHub , your DB —
    
    You MUST know :
        - What Open code can DO 
        - What Open code can SEE
        - How to stay SAFE
        
        & thats where Permissions & Tools comes into picture !





═══════════════════════════════════════════════
🎯 Section 1 : Understanding In-built Tools + MCP Tools 🛠️
═══════════════════════════════════════════════

First Understand what In-built Tools are , then we will understand with respect to External Tools (GitHub , your DB )



📌 What are Tools?

    Open Code internally uses different tools to complete tasks.

    Examples:

    Reading files → uses Read Tool
    Editing code → uses Edit Tool
    Running terminal commands → uses Bash Tool
    Searching internet → uses WebSearch Tool



    Above are Built-In Tools !
    Similarly , u have External Tools inside every MCP server !
    and whenever u give some prompt/task , Tools of that MCP server will be called & your task will be done !
    
    

    [ VS CODE ] : 
    name all the tools of connnected github mcp server !
    name all the tools of connnected context7 mcp server

    in short , An MCP server is a collection of tools..









    [Open Code logo]
   │
   ├── Built-in Tools
   │      Read
   │      Edit
   │      Bash
   │      WebSearch
   │
   ├── GitHub MCP
   │      44 GitHub tools
   │
   ├── SQLite MCP
   │      11 Database tools
   │
   └── Context7 MCP
          2 tools







You
   │
   ▼
"Check yesterday's Amazon sales"

   │
   ▼
Open Code understands the task

   │
   ▼
Chooses the correct MCP Server

   │
   ▼
Chooses the correct Tool

   │
   ▼
Runs the Tool

   │
   ▼
Returns the result to you



📌 What are Permissions?

    Permissions mean:

    You approve Open Code to use certain tools or commands.
                OR
    Claude Code automatically uses built-in or MCP-provided tools or prompts without your approval ..


    Example:
    Allow Opencode to use Edit Tool [ show logo]
    Allow Opencode  to access WebSearch[ show logo]
    Allow GITHUB MCP[ show logo] to access repo , fetch data from Github[ show logo]
    Allow Context7 MCP[ show logo] to fetch latest docs [ show logo]





'''
```
